```python

# ==============================
# COMPLETE EDA ON PENGUINS DATASET
# ALL CODE IN ONE CELL
# ==============================

# Import Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load Dataset
df = sns.load_dataset("penguins")

# ==============================
# BASIC DATASET INFORMATION
# ==============================

print("FIRST 5 ROWS")
print(df.head())

print("\nLAST 5 ROWS")
print(df.tail())

print("\nSHAPE OF DATASET")
print(df.shape)

print("\nCOLUMN NAMES")
print(df.columns)

print("\nDATA TYPES")
print(df.dtypes)

print("\nCOMPLETE INFORMATION")
print(df.info())

print("\nSTATISTICAL SUMMARY")
print(df.describe())

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE VALUES")
print(df.duplicated().sum())

print("\nUNIQUE SPECIES")
print(df["species"].unique())

print("\nSPECIES COUNT")
print(df["species"].value_counts())

# ==============================
# HANDLE MISSING VALUES
# ==============================

df = df.dropna()

# ==============================
# GROUPBY ANALYSIS
# ==============================

print("\nAVERAGE BODY MASS BY SPECIES")
print(df.groupby("species")["body_mass_g"].mean())

print("\nAVERAGE BILL LENGTH BY SEX")
print(df.groupby("sex")["bill_length_mm"].mean())

# ==============================
# FEATURE ENGINEERING
# ==============================

df["bill_ratio"] = df["bill_length_mm"] / df["bill_depth_mm"]

print("\nNEW COLUMN CREATED: bill_ratio")
print(df.head())

# ==============================
# CORRELATION MATRIX
# ==============================

print("\nCORRELATION MATRIX")
print(df.corr(numeric_only=True))

# ==============================
# VISUALIZATIONS
# ==============================

# 1. HISTOGRAM
plt.figure(figsize=(6,4))
plt.hist(df["body_mass_g"], bins=20)
plt.title("Histogram of Body Mass")
plt.xlabel("Body Mass")
plt.ylabel("Frequency")
plt.show()

# 2. COUNT PLOT
plt.figure(figsize=(6,4))
sns.countplot(x="species", data=df)
plt.title("Count Plot of Species")
plt.show()

# 3. BAR PLOT
plt.figure(figsize=(6,4))
sns.barplot(x="species", y="body_mass_g", data=df)
plt.title("Average Body Mass by Species")
plt.show()

# 4. SCATTER PLOT
plt.figure(figsize=(6,4))
plt.scatter(df["bill_length_mm"], df["body_mass_g"])
plt.title("Bill Length vs Body Mass")
plt.xlabel("Bill Length")
plt.ylabel("Body Mass")
plt.show()

# 5. BOX PLOT
plt.figure(figsize=(6,4))
sns.boxplot(x=df["body_mass_g"])
plt.title("Boxplot of Body Mass")
plt.show()

# 6. VIOLIN PLOT
plt.figure(figsize=(6,4))
sns.violinplot(x="species", y="body_mass_g", data=df)
plt.title("Violin Plot")
plt.show()

# 7. LINE PLOT
plt.figure(figsize=(6,4))
sns.lineplot(x="bill_length_mm", y="body_mass_g", data=df)
plt.title("Line Plot")
plt.show()

# 8. PIE CHART
plt.figure(figsize=(6,6))
df["species"].value_counts().plot.pie(autopct='%1.1f%%')
plt.title("Species Percentage")
plt.ylabel("")
plt.show()

# 9. KDE PLOT
plt.figure(figsize=(6,4))
sns.kdeplot(df["flipper_length_mm"], fill=True)
plt.title("KDE Plot of Flipper Length")
plt.show()

# 10. HEATMAP
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()

# 11. PAIRPLOT
sns.pairplot(df, hue="species")
plt.show()

# 12. STRIP PLOT
plt.figure(figsize=(6,4))
sns.stripplot(x="species", y="body_mass_g", data=df)
plt.title("Strip Plot")
plt.show()

# 13. SWARM PLOT
plt.figure(figsize=(6,4))
sns.swarmplot(x="species", y="bill_length_mm", data=df)
plt.title("Swarm Plot")
plt.show()

# 14. BOXPLOT WITH HUE
plt.figure(figsize=(6,4))
sns.boxplot(x="species", y="body_mass_g", hue="sex", data=df)
plt.title("Boxplot with Hue")
plt.show()

# 15. RELPLOT
sns.relplot(x="bill_length_mm",
            y="bill_depth_mm",
            hue="species",
            data=df)

plt.show()

# ==============================
# FINAL CONCLUSION
# ==============================

print("\nEDA COMPLETED SUCCESSFULLY")
print("Penguins dataset analyzed with multiple plots and statistics.")
```
